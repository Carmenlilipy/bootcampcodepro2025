Challenge 2: Adivina el Número Crea un juego simple en el que el usuario tiene que adivinar un número secreto. El usuario puede hacer 3 intentos. Cada vez que falle, el programa debe decir si el número secreto es mayor o menor al ingresado . Si el usuario adivina, debe mostrar "¡Felicidades! Adivinaste".

```python
#Crear un numero secreto a ser adivinado
import random 
adivinanza = random.randint(0, 15)
#intentos son 3 cantidad de veces a intentar
numero = 3 
print(adivinanza)
while numero >0:
   numero = numero-1
   numero_a_ingresar= int(input())
   if numero_a_ingresar > adivinanza:
    
      print("Numero ingresado es mayor")
   elif numero_a_ingresar == adivinanza:
      print("Eeeaaa ganasteeee!!")
      break
   else:
      print("Numero ingresado es menor")

```

Challenge 3: La Clase Animal Crea una clase llamada Animal que tenga los atributos: ● Nombre ● edad La clase debe tener un método llamado presentarse() que imprima algo como:

```python
#Crear una clase animal
class Animal:
    
#atributo de instancia
    def __init__(self, nombre, edad):
        self.nombre = nombre 
        self.edad = edad 
#metodo de presentacion
    def presentarse(self):
        print(f"Hola soy {self.nombre} y tengo {self.edad} anhos")
perro = Animal("Luna",2)
gato = Animal("Kira", 6)

# imprimir las acciones 
perro.presentarse()
gato.presentarse()
```

Challenge 4: Peluche Crea una clase Peluche con los atributos: ● nombre ● tamaño La clase debe tener un método abrazar() que imprima algo como: "¡Es hora de abrazar a [nombre], el peluche de tamaño [tamaño]!" Luego, crea un peluche y haz que sea abrazado. Conceptos usados: ● Clases, Atributos, Métodos.

```python
#Crear una clase peluche
class Peluche:
    
#atributo de instancia
    def __init__(self, nombre, tamanho):
        self.nombre = nombre 
        self.tamanho = tamanho
#metodo de presentacion
    def abrazar(self):
        print(f"Es hora de abrazar a {self.nombre} y mido {self.tamanho}")
winnie = Peluche("Carmen", 80)

# imprimir las acciones 
winnie.abrazar()
```

Challenge 5: Multiplicar una Lista de Números Escribe una función llamada multiplicar_lista() que reciba una lista de números como parámetro y devuelva el resultado de multiplicar todos los elementos de la lista.


```python
lista = [1, 5, 9, 7, 4]
#definir una lista a multiplicar
def multiplicar_lista(numeros):
    aux = 1    
    for i in numeros:
        aux = aux * i 
    return aux
multiplicar_lista(lista)



```

Challenge 6: Acumulador de Números Escribe una función llamada acumulador() que reciba un número n como parámetro y devuelva la suma de todos los números desde 1 hasta n.

```python
#definir funcion llamado acumulador
def acumulador(n):
    aux = 0
    for i in range(1, n):
        aux = aux + i
    return aux
acumulador(9)
```
