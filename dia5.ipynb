Challenge 2
Crear una tarjeta personal. El usuario (mediante input) puede elegir:

1 Crear datos nuevos
2 Modificar datos
3 Eliminar datos

```python
# Crear tarjeta
tarjeta = {"nombre": "", "email": "", "telefono": "", "direccion": "",}

def crear_datos():
    print("Ingrese los nuevos datos:")
    tarjeta["nombre"] = input("nombre")
    tarjeta["telefono"] = input("telefono:")
    tarjeta["email"] = input("Email:")
    tarjeta["direccion"] = input("Direccion:")
    print("Datos creados")
def mostrar_menu():
    print("Menu tarjeta personal")
    print("Crear datos nuevos")
    print("Modificar datos")
    print("Eliminar datos")
    print("Salir")

def mostrar_datos():
    print("Datos actuales:")
    for campo, valor in tarjeta.items():
        print(f"{campo.capitalize()}: {valor}")

def modificar_datos():
    if all(valor == "" for valor in tarjeta.values()):
        print("No hay datos para modificar.")
        return
    mostrar_datos()
    campo = input("¿Qué campo desea modificar? (nombre, telefono, email, direccion): ")
    if campo in tarjeta:
        nuevo_valor = input(f"Nuevo valor para {campo}: ")
        tarjeta[campo] = nuevo_valor
        print("Dato modificado exitosamente.")
    else:
        print("Campo no válido.")

def eliminar_datos():
    if all(valor == "" for valor in tarjeta.values()):
        print("No hay datos para eliminar.")
        return
    mostrar_datos()
    campo = input("¿Qué campo desea eliminar? (nombre, telefono, email, direccion): ")
    if campo in tarjeta:
        tarjeta[campo] = ""
        print("Dato eliminado exitosamente.")
    else:
        print("Campo no válido.")


# Bucle principal
while True:
    mostrar_menu()
    opcion = input("Seleccione una opción (1-4): ")

    if opcion == "1":
        crear_datos()
    elif opcion == "2":
        modificar_datos()
    elif opcion == "3":
        eliminar_datos()
    elif opcion == "4":
        print("¡Hasta luego!")
        break
    else:
        print("Opción no válida. Intente de nuevo.")

```

Challenge 

Crear dos metodos para nuestro animal y ejecutar, debe ser algo especifico de ese animal 
ejemplo "cuando hablo digo X ". ejemplo cuando me muevo, hago X


```python
# Definimos la clase gato
# atributos de clases comunes
class Gato:
    tipo = "felino"
    patas = 4
    orejas = 2
    cola = 1

    #atributo de instancia
    def __init__(self, color, raza):
        self.color = color 
        self.raza = raza
        
# metodo para describir acciones 
    def mueve(self):
        return ("menea la cola")
    def maulla(self):
        return("miaaauuu") 

# creamos un objeto de la clase gato 
kira = Gato

# ejecutamos los metodos
print(kira.moverse())
print(kira.maullar())
```

Challenge 
Definir una clase madre que herede a sus clases hijas dos atributos de instancia. mostrar ambos atributos en pantalla
observacion: Editar las clases hijas para que no se inicialicen.

```python
# Definición de la clase madre
class ClaseMadre:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

# Definición de la primera clase hija
class ClaseHija1(ClaseMadre):
    apodo = "Lili"


# Definición de la segunda clase hija
class ClaseHija2(ClaseMadre):
    apodo = "Chuchi"

# Crear instancias de las clases hijas
hija1 = ClaseHija1("Liliana",35 )
hija2 = ClaseHija2("Carlos", 20)

# Mostrar los atributos en pantalla
print(f"ClaseHija1 - Nombre: {hija1.nombre}, Edad: {hija1.edad}")
print(f"ClaseHija2 - Nombre: {hija2.nombre}, Edad: {hija2.edad}")
```
